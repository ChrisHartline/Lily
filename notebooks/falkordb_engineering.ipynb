```python
# Configuration - Enter your FalkorDB Cloud credentials
FALKOR_HOST = "your-instance.cloud"  # @param {type:"string"}
FALKOR_PORT = 55041  # @param {type:"integer"}
FALKOR_USERNAME = "falkordb"  # @param {type:"string"}
FALKOR_PASSWORD = ""  # @param {type:"string"}
FALKOR_GRAPH = "claralilymem"  # @param {type:"string"}
```

```python
# Test FalkorDB connection
from falkordb import FalkorDB

db = FalkorDB(
    host=FALKOR_HOST,
    port=FALKOR_PORT,
    username=FALKOR_USERNAME,
    password=FALKOR_PASSWORD
)

graph = db.select_graph(FALKOR_GRAPH)
print("✅ FalkorDB connected!")

# Test query - count nodes
result = graph.query("MATCH (n) RETURN count(n) as total")
print(f"Total nodes: {result.result_set[0][0]}")
```

```python
# Sample Cypher queries for Clara's memory graph

# Create an entity
graph.query("""
MERGE (e:Entity {name: 'Chris', type: 'person'})
SET e.importance = 1.0
RETURN e
""")

# Create a relationship
graph.query("""
MATCH (a:Entity {name: 'Chris'})
MATCH (b:Entity {name: 'Clara'})
MERGE (a)-[r:KNOWS]->(b)
SET r.strength = 1.0
RETURN a, r, b
""")

# Query relationships
result = graph.query("""
MATCH (a:Entity)-[r]->(b:Entity)
RETURN a.name, type(r), b.name
LIMIT 10
""")

for row in result.result_set:
    print(f"{row[0]} --[{row[1]}]--> {row[2]}")
```
